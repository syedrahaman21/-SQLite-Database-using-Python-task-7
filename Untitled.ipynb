```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 1: Create a new SQLite DB and connect
conn = sqlite3.connect("sales_data.db")
cursor = conn.cursor()
```

```python
# Step 2: Create a sales table
cursor.execute('''
    CREATE TABLE IF NOT EXISTS sales (
        id INTEGER PRIMARY KEY,
        product TEXT,
        quantity INTEGER,
        price REAL
    )
''')
```

```python
# Step 3: Insert some dummy data
sample_data = [
    ("Apple", 10, 1.5),
    ("Banana", 5, 0.5),
    ("Orange", 8, 1.0),
    ("Apple", 4, 1.5),
    ("Banana", 7, 0.5),
    ("Orange", 6, 1.0)
]
cursor.executemany("INSERT INTO sales (product, quantity, price) VALUES (?, ?, ?)", sample_data)
conn.commit()
```

```python
# Step 4: Run SQL query to get summary
query = '''
    SELECT 
        product, 
        SUM(quantity) AS total_qty, 
        SUM(quantity * price) AS revenue 
    FROM sales 
    GROUP BY product
'''
df = pd.read_sql_query(query, conn)
```

```python
# Step 5: Print the summary
print("Sales Summary:")
print(df)
```

```python
# Step 6: Plot a simple bar chart of revenue by product
df.plot(kind='bar', x='product', y='revenue', legend=False, color='skyblue')
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.tight_layout()
plt.savefig("sales_chart.png")
plt.show()
```
